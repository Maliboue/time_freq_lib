```python
import matplotlib.pyplot as plt
import numpy as np
from scipy.signal import periodogram, welch
```

## Make data

```python
nt = 1000
N = 1000
dt = 1

signals = np.random.randn(N, nt)
tvec = np.arange(nt)*dt
```

## Periodogram

```python
fs, spectra = periodogram(signals, fs=1/dt)
```

```python
fs = fs[1:-1]
spectra = spectra[:, 1:-1]
```

```python
f,ax = plt.subplots()
ax.plot(fs, spectra.mean(axis=0))
```

```python
plt.hist(spectra.ravel(), bins=50);
plt.yscale('log')
```

## Welch

```python
fs, spectra = welch(signals, fs=1/dt, nperseg=50, noverlap=0, window='boxcar')
```

```python
fs = fs[1:-1]
spectra = spectra[:, 1:-1]
```

```python
f,ax = plt.subplots()
ax.plot(fs, spectra.mean(axis=0))
```

```python
plt.hist(spectra.ravel(), bins=50);
# plt.yscale('log')
```
