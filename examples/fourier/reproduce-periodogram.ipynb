```python
import matplotlib.pyplot as plt
import numpy as np
from scipy.signal import periodogram, welch
```

## Make data

```python
nt = 20
dt = 0.02

signal = np.random.randn(nt)
tvec = np.arange(nt)*dt

fs = 1/dt

fnyq = fs/2
df = 1/(nt*dt)
```

## Periodogram

```python
x_ft = np.fft.fft(signal)

x_pow = abs(x_ft)**2
freqs = np.fft.fftfreq(nt, dt)

sele = freqs>=0

x_pow_oneside = x_pow[sele]*2
freqs_oneside = freqs[sele]

x_pow_oneside = x_pow_oneside*dt/nt
```

```python
plt.plot(freqs, x_pow)
```

```python
fs, sp = periodogram(signal, fs=1/dt)
```

```python
f,ax = plt.subplots()
ax.plot(fs, sp)
plt.plot(freqs_oneside, x_pow_oneside, marker='o')
```
