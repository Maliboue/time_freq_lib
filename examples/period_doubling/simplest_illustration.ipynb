## Illustrate how signal and it's spectrum look like

```python
import numpy as np
import matplotlib.pyplot as plt
from sys import path
path.append('../../signals/')

import signals as sg
```

```python
from scipy.signal import periodogram
```

```python
dur = 600
freq = 0.03
fs = 2
t, x = sg.cosine(dur, freq, fs)

x_double = x + sg.cosine(dur, freq/2, fs, phase=np.pi/2)[1]
```

```python
frq, p = periodogram(x, fs=fs)
frq, p_double = periodogram(x_double, fs=fs)
```

```python
f, (ax,ax1) = plt.subplots(2,1,figsize=(10,9))

ax.plot(t, x)
ax1.step(frq, p, where='mid')
ax1.set_xlim(0, 0.05)
ax.set(title='Signal', xlabel='Time (s)')
ax1.set(title='Spectrum', xlabel='Frequency (Hz)')
```

```python
f, (ax,ax1) = plt.subplots(2,1,figsize=(10,9))

ax.plot(t, x_double)
ax1.step(frq, p_double, where='mid')
ax1.set_xlim(0, 0.05)
ax.set(title='Signal', xlabel='Time (s)')
ax1.set(title='Spectrum', xlabel='Frequency (Hz)')
```
